- z = 1/(1+e^(-z))
##### This convert values from : + inf to - Inf into ==== 0 to 1

#### 8. End to End Logistic Regression

```python
# Student Pass or fail Prediction
import pandas as pd
import numpy as np
data = {
    'Hours_Studied': [2, 3, 5, 1, 4, 6, 7, 8],
    'Passed': [0, 0, 1, 0, 1, 1, 1, 1]
}

df=pd.DataFrame(data)
df
```

```python
# Seperating Independent and Dependent variable
x= df[["Hours_Studied"]]
y= df[["Passed"]]
print(x)
print(y)
```

```python
# Train Test Split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train)
```

```python
print("Training data size: ",len(x_train))
print("Testing data size: ",len(x_test))
```

```python
# Train Logistic Regression Model
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
m= model.coef_[0][0]
c=model.intercept_[0]
print("m: ",m)
print("c: ",c)
```

```python
x=5
z=m*x+c
print("z: ",z)
sigmoid = 1/(1+math.exp(-z))
print("Sigmoid value: ",sigmoid)
probability=sigmoid
print("Probability of passing: ",probability)
if probability > 0.6:
    print("Prediction  : Pass (Class 1)")   
else :
    print("Prediction  : Fail (Class 0)")
```

```python
pred=model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
c=confusion_matrix(y_test,pred)
print("Confusion Matrix: \n",c)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,classification_report
accuracy = accuracy_score(y_test,pred)
precision = precision_score(y_test,pred)
recall = recall_score(y_test,pred)
f1 = f1_score(y_test,pred)
clas=classification_report(y_test,pred)
print("Accuracy: ",accuracy)
print("Precision: ",precision)
print("Recall: ",recall)
print("F1 Score: ",f1)
print("Classification Report: \n",clas)
```
